```python
class Employee: #Parent Class
  def __init__(self,name:str,age:int):
    self.name=name
    self.age=age
  def getDetails(self):
    print(f"Hi My Name is {self.name} and My Age is {self.age}")

class Manager(Employee): # Child Class
  def __init__(self, name: str, age: int, color: str):
    super().__init__(name, age)   # call parent constructor
    self.color = color
  def managerDetails(self):
    print(f"My color is {self.color}  ")
```

```python
obj = Employee("Madan", 24)
obj.getDetails()
```

```python
obj2 = Manager("Madan", 24, "Red")
#obj2.getDetails()
obj2.managerDetails()
```

```python
class Parent1:
  def get_parent1(self):
    print("Hello, I am Parent 1")
class Parent2:
  def get_parent2(self):
    print("Hello, I am Parent 2")
class Child(Parent1,Parent2):
  def get_child(self):
    print("Hello, I am Child")
```

```python
obj3=Child()
obj3.get_child()
obj3.get_parent1()
obj3.get_parent2()
```

```python
class Grandpa:
  def get_Grandpa(self,name:str):
    self.name=name
    print(f"Hello, I am Grandfather My name is {self.name}")
class Father(Grandpa):
  def get_Father(self):
    print("Hello, I am Parent 2")
class Child(Father):
  def get_child(self):
    print("Hello, I am Child")
```

```python
obj4=Child()
obj4.get_child()
obj4.get_Grandpa("Ram")
obj4.get_Father()
```

```python
class Family:
  def __init__(self,name:str,age:int):
    self.name=name
    self.age=age
    print(f"Hello, My Name is {self.name} and My Age is {self.age}")
class Child1(Family):
  def __init__(self, name: str, age: int, color: str):
    super().__init__(name, age)
    self.color = color
  def child1Details(self):
    print(f"I am first child {self.color}")
class Child2(Family):
  def __init__(self, name: str, age: int, color: str):
    super().__init__(name, age)
    self.color = color
  def child2Details(self):
    print(f"My color is {self.color}")
```

```python
child=Child1("Madan",24,"Red")
child.child1Details()
child
```

```python
class Father:
    def __init__(self, father_name):
        self.father_name = father_name

    def father_details(self):
        print(f"Father Name: {self.father_name}")


class Child1(Father):
    def __init__(self, father_name, child1_name, age):
        super().__init__(father_name)
        self.child1_name = child1_name
        self.age = age

    def child1_details(self):
        print(f"Child1 Name: {self.child1_name}")
        print(f"Age: {self.age}")


class Child2(Father):
    def __init__(self, father_name, child2_name, color):
        super().__init__(father_name)
        self.child2_name = child2_name
        self.color = color

    def child2_details(self):
        print(f"Child2 Name: {self.child2_name}")
        print(f"Favorite Color: {self.color}")
```

```python
c1 = Child1("Ram", "Madan", 24)
c1.father_details()
c1.child1_details()

print("------------------------")

c2 = Child2("Ram", "Sita", "Red")
c2.father_details()
c2.child2_details()
```
