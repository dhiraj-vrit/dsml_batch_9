```python
# method 1 kid level
hislist = [1,2,3,4,5,6,7,8,8,9]
herlist = []
for i, ele in enumerate(hislist, 1):
  a = ele * 2
  herlist.append(a)
print(herlist)
```

```python
# method 2 more efficient and interview level
hislist = [1,2,3,4,5,6,7,8,8,9]
for i, ele in enumerate(hislist):
  hislist[i] = ele * 2
print(hislist)
```

```python

# method 3 easy lvel

his = [1,2,3,4,5,6,7,8,8,9]
i = 0
for ele in his:
  hislist[i] = ele * 2
  print(i)
  i += 1
print(hislist)
```

```python
names = ['purna', 'singh', 'taman', 'hello', 'kiran']
grade = [2 ,3, 4, 6, 8]
mark = ['A+', 'A-' , 'C+', 'B+', 'B-']
print(type(zip(names, grade, mark)))
newlist = list(zip(names, grade, mark))
for i , (a,b,c) in enumerate(newlist):
  print(i, a, b, c)
```

```python
def my_function(**kid):
  print("His last name is " + kid["fname"])

my_function(fname = "Tobias", lname = "Refsnes")
```

```python
def my_function(*kids):
  print("The youngest child is " + kids[2])

my_function("Emil", "Tobias", "Linus")
```

```python
def my_function(*args):
  print("Type:", type(args))
  print("First argument:", args[0])
  print("Second argument:", args[1])
  print("All arguments:", args)

my_function("Emil", "Tobias", "Linus")
```

```python
def my_function(greeting, *names):
  for name in names:
    print(greeting, name)

my_function("Hello", "Emil", "Tobias", "Linus")
```

```python
names = ["pur", "ran", "esh", "hel", "kir"]
address = ["ktm", "lalitpur", "bhaktaput", "ktm", "kkk"]
for na, add in zip(names, address):
  print(f"name : {na}  | address : {add}")
```

```python
def my_function(*numbers):
  total = 0
  for num in numbers:
    total += num
  return total

print(my_function(1, 2, 3))
print(my_function(10, 20, 30, 40))
print(my_function(5))
```

```python
def my_function(title, *args, **kwargs):
  print("Title:", title)
  print("Positional arguments:", args)
  print("Keyword arguments:", kwargs)

my_function("User Info", "Emil", "Tobias", age = 25, city = "Oslo")
```
