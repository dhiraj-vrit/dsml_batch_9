```python
class A:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def get_details(self):
        print("Class A Details")
        print("Name:", self.name)
        print("Age:", self.age)


class B:
    def get_details(self, obj):
        print("Class B Accessing Class A Details")
        print("Name:", obj.name)
        print("Age:", obj.age)


# Create object of Class A
a1 = A("Madan", 25)

# Call method from Class A
a1.get_details()

print("--------------------")

# Create object of Class B
b1 = B()

# Pass object of A to B
b1.get_details(a1)
```

```python
from abc import ABC, abstractmethod

# Abstract Base Class
class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass


# Child Class 1
class EsewaPayment(Payment):
    def pay(self, amount):
        print(f"Paid {amount} using Esewa")


# Child Class 2
class KhaltiPayment(Payment):
    def pay(self, amount):
        print(f"Paid {amount} using Khalti")
```

```python
esewa = EsewaPayment()
esewa.pay(1000)

print("-------------------")

khalti = KhaltiPayment()
khalti.pay(250)
```

```python
from abc import ABC, abstractmethod

# Abstract Class
class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass


# Child Class
class Rectangle(Shape):

    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)


```

```python
# Object Creation
rect = Rectangle(10, 5)

print("Area:", rect.area())
print("Perimeter:", rect.perimeter())
```

```python
class Point:
    def __init__(self, x, y, z):
        self.x = x
        self.y = y
        self.z = z

    def __repr__(self):
        return f"Point(x={self.x}, y={self.y}, z={self.z})"

    def __sub__(self, other):
        x = self.x - other.x
        y = self.y - other.y
        z = self.z - other.z
        return Point(x, y, z)
```

```python
p1 = Point(5, 6, 7)
p2 = Point(1, 2, 3)

result = p1 - p2
print(result)
```

```python
class Point:
  def __init__(self,x,y,z):
    self.x=x
    self.y=y
    self.z=z
  def __repr__(self):
    return f"Point(x={self.x},y={self.y},z={self.z})"
  def __mul__(self,other):
    x=self.x* other.x
    y=self.y* other.y
    z=self.z* other.z
    return Point(x,y,z)
```

```python
p1 = Point(2, 3, 4)
p2 = Point(5, 6, 7)

print(p1 * p2)
```

```python
#Duck Typing
class Car:
  def drive(self):
    print("Driving a car")
class Bike:
  def drive(self):
    print("Riding a bike")

```

```python
def start_driving(vehicle):
  vehicle.drive()
```

```python
car=Car()
bike=Bike()
start_driving(car)
start_driving(bike)
```

```python
def flying_test(bird):
  bird.fly()
```
