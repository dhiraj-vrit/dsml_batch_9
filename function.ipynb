```python
# recursive function
def factorial(a):
  if (a == 1):
    return 1
  return a * factorial(a-1)

print(factorial(7))
```

```python
# finction with default parameter
def  purna(a, b=1):
  sum = a + b
  return sum
print(purna(5))
```
