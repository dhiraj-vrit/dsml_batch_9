# Shallow Copy
In shallow copy it is linked and it copy even after the list is append.

```python

second_data = [[1, 2], [3, 4]]

secondfinal_data = copy.copy(second_data)

second_data[0].append(99)

print("Original:", second_data)
print("Shallow Copy:", secondfinal_data)
```
