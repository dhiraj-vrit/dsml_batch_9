```python
first1 = {
    "name" : "apple pvt ltd",
    "founder" : "purna lama" , 
}
second1 = {
    "type" : "charity",
    "age" : 8,
    "price" : 0.01,
    "work_style" : "remote"
}
third1 = {
    "work_station" : "nepal",
     "avg_salary" : 10000
}
fith = (first1 | second1 | third1) # can also be 
print(fith)
four1 = {**first1 , **second1, **third1} # **kwargs
print(four1)
```
