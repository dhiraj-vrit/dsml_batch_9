```python
sum=lambda a,b,c: a + b-c
sum(5,3,6)
```

```python
check = lambda a, b: "a is Greater" if a > b else "b is Greater"

print(check(10, 7))
print(check(5, 9))
```

```python
check = lambda a, b, c: (
    "a is Greater" if a > b and a > c else
    "b is Greater" if b > a and b > c else
    "c is Greater"
)

print(check(10, 7, 5))
print(check(5, 9, 2))
print(check(4, 4, 4))
```

```python
check1 = lambda a: print("Even",a) if a % 2 == 0 else print("Odd",a)
check1(10)
check1(5)
```
