```python
class person:
  def __init__(self,name:str,age:int="25"):
    self.name=name
    self.age=age
  def get_details(self):
    print(f"name is {self.name} and I am {self.age} years old")
```

```python
p1=person("ram",20)
p1.get_details()
```

```python
p2=person("shyam")
p2.get_details()

```

```python
class Telivision:
  def __init__ (self,name:str,model:str,company:str,price:int,size:int,Types:str):
    self.name=name
    self.model=model
    self.company=company
    self.price=price
    self.size=size
    self.Types=Types
  def getquote(self):
    print("---------------------------------------------------------")
    print("This is your Confirmation email of Online Shopping ")
    print("---------------------------------------------------------")
    print(f"Mr.{self.name} You Have order {self.size} inch {self.Types} Tv of model no {self.model} Tv of the Company {self.company} Which is price at {self.price} ")
```

```python
Tv1=Telivision("Madan Khanal","0065AUHD","Samsung","600$","65","Oled")
Tv1.getquote()
```

```python
class Trip:
  def __init__(self,name:str,weight:float,height:float,destination:str,miles:int,):
    self.name = name
    self.weight = weight
    self.height = height
    self.destination = destination
    self.miles = miles
  def yourtrip(self):
    print(f"Hi {self.name} of weight {self.weight} and height {self.height} feet \nYour Destination is {self.destination} Which is {self.miles} m")
```

```python
T1=Trip("Raju Rastogi","68.9","5.8","Dubai","8848")
T1.yourtrip()
```
