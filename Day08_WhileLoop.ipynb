```python
n = int(input("Enter a number: "))

# Keep looping until n is a single digit (0-9)
while n >= 10:
    total = 0

    # This loop breaks the number down and adds the digits
    while n > 0:
        total += n % 10  # Get the last digit (e.g., 55 -> 5)
        n = n // 10      # Remove the last digit (e.g., 55 -> 5)

    n = total            # Update n with the new sum
    print(f"Current sum: {n}")

print(f"Final digit: {n}")
```

```python
prime=int(input("enter the number: "))
for i in range(2,prime):
  if prime % i ==0:
    print(f"{prime} is not a prime number")
    break
else:
  print(f"{prime} is a prime number")
```
